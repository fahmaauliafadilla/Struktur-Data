```python
def coin_change_greedy (amount, coins):
  coins.sort(reverse=True)
  result = []
  for coin in coins:
    while amount >= coin:
      amount -= coin
      result.append (coin)
  return result

# Contoh penggunaan
amount = 57
coins = [25, 10, 5, 1]
change = coin_change_greedy (amount, coins)
print("Koin yang digunakan:", change)
```

```python
def fractional_knapsack(items, capacity):

# items list of tuples (value, weight)
  items = sorted(items, key=lambda x: x[0]/x [1], reverse=True)
  total_value = 0.0
  for value, weight in items:
    if capacity >= weight:
      capacity -= weight
      total_value += value
    else:
      total_value += value * (capacity / weight)
      break
  return total_value

#Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)] # (nilai, berat)
capacity = 50
max_value = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
```

```python
import heapq

def prim_mast (graph, start):
  visited = set ()
  min_heap = [(0, start, None)]
  total_weight = 0

  while min_heap:
    weight, node, _ = heapq.heappop(min_heap)

    if node in visited:
      continue

    visited.add(node)
    total_weight += weight

    for neighbor, edge_weight in graph[node]:
      if neighbor not in visited:
        heapq.heappush(min_heap, (edge_weight, neighbor, node))

  return total_weight

#Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}

print("Total bobot MST (Prim):", prim_mast (graph, 'A'))
```

```python
def kruskal_mst (edges, n_nodes):
    # Initialize parent array for Disjoint Set Union (DSU)
    # Each node is initially its own parent
    parent = {i: i for i in range(n_nodes)}

    # The 'find' operation with path compression
    def find(x):
        if parent[x] == x:
            return x
        parent[x] = find(parent[x]) # Path compression
        return parent[x]

    # The 'union' operation by rank or size (here simple union by setting one root as parent of another)
    def union(x, y):
        root_x = find(x)
        root_y = find(y)

        if root_x != root_y:
            parent[root_y] = root_x
            return True # Successfully united
        return False # Already in the same set

    # Sort edges by weight in non-decreasing order
    edges.sort(key=lambda x: x[2])

    total_weight = 0
    num_edges_in_mst = 0

    # Iterate through sorted edges
    for u, v, weight in edges:
        # If adding this edge does not form a cycle (i.e., u and v are in different sets)
        if union(u, v):
            total_weight += weight
            num_edges_in_mst += 1
            # An MST for n_nodes has exactly n_nodes - 1 edges
            if num_edges_in_mst == n_nodes - 1:
                break # MST is complete

    return total_weight

# Example usage:
# edges: (node1, node2, weight), nodes are represented by integers (0-indexed)
edges_example = [(0, 1, 2), (1, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
n_nodes_example = 4 # Number of nodes (0, 1, 2, 3)

print("Total bobot MST (Kruskal):", kruskal_mst(edges_example, n_nodes_example))
```
