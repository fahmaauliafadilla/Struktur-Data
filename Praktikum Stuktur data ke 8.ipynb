```python
# Membuat  graf tak berarah menggunakan adjcency  list
graph = {
      'A':['B','C'],
      'B':['A','D','E'],
      'C':['A','F'],
      'D':['B'],
      'E':['B','F'],
      'F':['C','E']
}

# Menampilkan graf
def print_graph(graf):
  for node in graph:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
from collections import deque


def bfs(graph, start):
    visited = set()
    queue = deque([start])


    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])


# Contoh pemanggilan
bfs(graph, 'A')
```

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()


    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)


# Contoh pemanggilan
dfs(graph, 'A')
```

```python
# Lengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graf

def count_neighbors(graph):
    result = {}

    for node in graph:
        # Menghitung jumlah tetangga
        result[node] = len(graph[node])

    return result


# Contoh graf
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "D"],
    "D": ["B", "C"],
    "E": ["B"]
}

# Menampilkan hasil
print(count_neighbors(graph))
```

```python
# Tugas
# Buatlah graf tak berarah baru dengan minimal 5 simpul dan 7 sisi, lalu tampilkan dengan fungsi print_graph.

class Graph:
    def __init__(self):
        self.graph = {}

    def add_edge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)
        self.graph[v].append(u)

    def print_graph(self):
        print("Graf:")
        for node in self.graph:
            print(f"{node}, -> {self.graph[node]}")

# Membuat graf baru
g = Graph()

# Minimal 5 simpul dan 7 sisi
g.add_edge("A", "B")
g.add_edge("A", "C")
g.add_edge("B", "D")
g.add_edge("C", "D")
g.add_edge("C", "E")
g.add_edge("D", "E")
g.add_edge("B", "E")

# Menampilkan graf
g.print_graph()
```

```python
# Implementasikan BFS dan DFS dari simpul awal pilihan Anda.
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


def bfs(graph, start):
    visited = set()
    queue = [start]

    visited.add(start)

    while queue:
        node = queue.pop(0)
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


print("\nDFS dari A:")
dfs(g.graph, "A")

print("\nBFS dari A:")
bfs(g.graph, "A")
```

```python
# Modifikasi fungsi BFS agar mengembalikan list urutan kunjungan, bukan hanya mencetak.
def bfs_list(graph, start):
    visited = set()
    queue = [start]
    hasil = []

    visited.add(start)

    while queue:
        node = queue.pop(0)
        hasil.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return hasil


print("\nBFS dalam bentuk list:")
print(bfs_list(g.graph, "A"))
```

```python
# Buat fungsi find_path(graph, start, end) yang mengembalikan satu jalur dari start ke end jika ada, menggunakan DFS.
def find_path(graph, start, end, path=[]):
    path = path + [start]

    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)

            if new_path:
                return new_path

    return None


print("\nPath dari A ke E:")
print(find_path(g.graph, "A", "E"))
```

```python
# Tambahkan fungsi is_connected(graph) untuk mengecek apakah semua simpul dalam graf saling terhubung (gunakan kombinasi DFS + logika).
def is_connected(graph):
    nodes = list(graph.keys())

    if not nodes:
        return True

    visited = set()

    def dfs_connected(node):
        visited.add(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_connected(neighbor)

    dfs_connected(nodes[0])

    return len(visited) == len(nodes)


print("\nApakah graf terhubung?")
print(is_connected(g.graph))
```
